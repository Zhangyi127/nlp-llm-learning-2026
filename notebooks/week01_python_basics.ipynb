```python
def count_words(text):
    """统计英文句子单词数量，按空格分割"""
    word_list = text.split()
    return len(word_list)

# 测试
sentence = "Large language models change the world"
num = count_words(sentence)
print(f"句子词数：{num}")
```

```python
# 故意出错代码
num = 12345
result = num.split()
```

split()是字符串 str 的方法；变量 num 是整数 int 类型，整数没有 split 这个函数，调用就报错。
修正方式：把数字转为字符串 str(num).split()。
